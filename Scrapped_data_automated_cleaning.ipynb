```python
import pandas as pd
# reading csv file
df = pd.read_csv(r"C:\Users\User\Downloads\Course Materials\Mobiles_scrapped.csv")
```

```python
# stripping extra characters from rating 
df['Rating'] = df['Rating'].str.strip('%;')
# cutting it to be 00.00 format
df['Rating'] = df['Rating'].str.slice(stop=5)
# changing periods to comas
df['Rating'] = df['Rating'].str.replace('.', ',')

# remove space in price
df['Price'] = df['Price'].str.split().str.join('')

# make company name in lower case for better normalization
df['Company'] = df['Company'].str.lower()

# function that takes mask condiotion to find rows where company needs to be replaced with first word in product name, it removes from product move
def company_to_first_name_word(mask):
    df.loc[mask, 'Company'] = (df.loc[mask, 'Product Name'].str.extract(r'^([^ ]+)', expand=False))
    df.loc[mask, 'Product Name'] = (df.loc[mask, 'Product Name'].str.replace(r'^\S+\s*', '', regex=True))

# function that takes mask condiotion to find rows where company needs to be replaced with second word in product name, it removes from product move
def company_to_second_name_word(mask):
    df.loc[mask, 'Product Name'] = (df.loc[mask, 'Product Name'].str.replace(r'^\S+\s*', '', regex=True))
    df.loc[mask, 'Company'] = (df.loc[mask, 'Product Name'].str.extract(r'^([^ ]+)', expand=False))
    df.loc[mask, 'Product Name'] = (df.loc[mask, 'Product Name'].str.replace(r'^\S+\s*', '', regex=True))
```
